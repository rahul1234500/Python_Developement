```python
# Break 

for number in range(1,10):
    if number == 4:
        break
    print(number)
```

```python
# continue 

for number in range(1,10):
    if number == 5:
        continue
    print(number)
```

```python
# Pass 

for row in range(1, 4):
    for column in range(1, 3):
        print(f"Row {row}, Column {column}")
```

```python
# return 

def find_age(age_list, target):
    for age in age_list:
        if age == target:
            return age
age_list = [12, 20, 18, 11]
result = find_age(age_list,18)
print(result)
```

```python
# comprehensions

list_1 = ["Rahul", "Amit", "Sarthak", "Manthan", "Ritik"]

list_2 = []

for i in list_1:
    
    list_2.append(len(i))
print(list_2)
```

```python
list_1 = ["Rahul", "Amit", "Sarthak", "Manthan", "Ritik"] 

list_2 = [len(i) for i in list_1]

print(list_2)
```

```python
list_1 = ["Rahul", "Amit", "Sarthak", "Manthan", "Ritik"]

d = {i : len(i) for i in list_1 }

print(d)
```
