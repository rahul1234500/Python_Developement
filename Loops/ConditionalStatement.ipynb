```python
# if block 
age = 22

if age >= 18:
    print("Eligible to apply")
```

```python
# if-else block 

stock =  0

if stock > 0 :
    print("Product available")
else :
    print("Out of Stock")
```

```python
# if-else-if
marks = 76

if marks >= 90:
    grade = "A"
elif marks >= 75:
    grade = "B"
elif marks >= 60:
    grade = "C"
else:
    grade = "D"

print(grade)
```

```python
# nested if-else

username = "rahul"
password_correct = True
account_active = True

if username:
    if password_correct:
        if account_active:
            print("Login successful")
        else:
            print("Account is inactive")
    else:
        print("Invalid password")
else:
    print("Username is required")
```
