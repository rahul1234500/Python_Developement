```python
# Combining so data frame 
# join operations  
# first step to create two different dataframes 
# with one same column name that is my access point and act as primary key like sql


# Employee Database consist only employee id and name 
df1 = pd.DataFrame(
    {
        "emp_id" : [1001,1002,1003,1004,1005],
        "emp_name" : ["Rahul", "Sarthak", "Pranay", "Aman", "Vedant"],
    }
)

# Department Database consist only employee id with salary 

df2 = pd.DataFrame(
    {
        "emp_id" : [1001,1006,1007,1008,1005],
        "emp_salary" : [500000,450000,350000,200000,100000],
    }
)

print("df1 \n",df1)
```

```python
print("df1 \n",df2)
```

```python
# inner join

inner_join = pd.merge(df1, df2, on="emp_id", how='inner')
```

```python
inner_join
```

```python
left_join = pd.merge(df1,df2, on="emp_id", how='left') 
```

```python
left_join
```

```python
right_join = pd.merge(df1,df2, on="emp_id", how='right')
```

```python
right_join
```

```python
# outer join

outer_join = pd.merge(df1,df2,on='emp_id',how='outer')
```

```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Rahul', 'Priya', 'Amit', 'Sneha'],
    'Age': [25, 30, 22, 28],
    'City': ['Pune', 'Mumbai', 'Delhi', 'Bangalore']
}, index=['a', 'b', 'c', 'd'])   # index labels a, b, c, d

df
```

**1. loc (Label based)**
Label (naam) se select karta hai.
Row ya column ka actual naam use karo.
Inclusive hota hai (start aur end dono include).

```python
df
```

```python
# Single row by label
df.loc['b']                 # Priya wali row

     # Age 25 se zyada wale
```

```python
# Multiple rows
df.loc[['a', 'c']]          # Rahul aur Amit

# Rows + specific columns
```

```python
df.loc['a':'c', ['Name', 'Age']]   # a se c tak, sirf Name aur Age
```

```python

# Condition ke saath
df.loc[df['Age'] >= 25, ['Name']] 
```

**2. iloc (Integer position based)**
Position (number) se select karta hai.
0, 1, 2... jaisa indexing (jaise list me).
Exclusive hota hai end ke liye (Python slicing jaisa).

```python
# First row (position 0)
df.iloc[:1]

```

```python

# 2nd aur 4th row (positions 1 aur 3)
df.iloc[[1, 3]]

```

```python
# Rows 0 se 2 tak (0,1,2) + columns 0 aur 1
df.iloc[0:3, 0:2] # [Rows, Columns]

```

```python
# Last row
df.iloc[-1]
```
