```python
# for loop 

skills = ["Python", "SQL", "Django"]
for skill in skills:
    print(skill)
```

```python
# with range function


for number in range(1,20,2):
    print(number)
```

```python
for number in range(10):
    print(number)
```

```python
for number in range(5,1,-1):
    print(number)
```

```python
amounts = [100, 250, 150]
total = 0

for amount in amounts:
    total+= amount

print(total)
```

```python
# While loop


attempt = 1

while attempt <= 3:
    print("Attempt : ", attempt)
    attempt += 1
```

```python
for row in range(1, 4):
    for column in range(1, 3):
        print(f"Row {row}, Column {column}")
```

```python
# Iteration through dictionary 

employeeData = {
   "id" : 101,
   "name" : "Rahul Kale",
   "position" : "Software Developer",
   "address" : "Pune, Maharshtra"
}

for key, value in employeeData.items():
    print(f"{key} : {value}")

```
