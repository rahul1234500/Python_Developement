```python
import math

class Circle:
    def __init__(self, radius: float):
        if radius < 0:
            raise ValueError("Radius cannot be negative.")
        self.radius = radius

    def area(self) -> float:
        """Calculate and return the area of the circle."""
        return math.pi * (self.radius ** 2)

    def perimeter(self) -> float:
        """Calculate and return the circumference."""
        return 2 * math.pi * self.radius

# Example usage:
my_circle = Circle(5)

print(f"Radius: {my_circle.radius}")
print(f"Area: {my_circle.area():.2f}")
print(f"Circumference: {my_circle.perimeter():.2f}")
```

```python
# instance variable and class variable 

class Employee :

    company_name = "TechCrop"
    employee_count = 0

    def __init__(self, name: str, salary: float):
        self.name = name
        self.salary = salary


        Employee.employee_count += 1

    def get_employee_name(self) -> str:
        print(f"EMPLOYEE NAME :\t {self.name}")

    def get_employee_salary(self) -> float:
        print(f"EMPLOYEE SALARY :\t {self.salary}")

    def get_employee_card(self):
        print(f"EMPLOYEE INFO ID-CARD \n{self.company_name}\n{self.name}\n{self.salary}\n{self.employee_count}")


emp1 = Employee("Alice", 85000)
emp2 = Employee("Rohan", 80000)
emp1.get_employee_card()
emp2.get_employee_card()
# 2. Accessing class variables (shared)
print(f"\nTotal employees: {Employee.employee_count}")
print(f"Via instance (emp1): {emp1.company_name}")
print(f"Via class: {Employee.company_name}")

# 3. Changing the class variable affects all instances
Employee.company_name = "TechCorp Global"
print(f"\nAfter rebrand:")
print(f"emp1 company: {emp1.company_name}")
print(f"emp2 company: {emp2.company_name}")

        
```

```python
# adding a method to class 

class Triangle :
    def __init__(self, base: float, height: float):
        self.base = base
        self.height = height

    def area_of_triangle(self) -> float:
        """Area of Triangle :"""
        return 1/2 * (self.base * self.height)


triangle = Triangle(10,20)

result = triangle.area_of_triangle()

print(f"Area of Triangle :{result}")      
```

```python
class BankAccount:
    def __init__(self, owner: str, balance: float, pin: int):
        self.owner = owner          # Public
        self._balance = balance     # Protected
        self.__pin = pin            # Private

    # Getter
    def get_balance(self, pin: int):
        if pin == self.__pin:
            return self._balance
        raise ValueError("Invalid PIN.")

    # Setter with validation
    def deposit(self, amount: float):
        if amount > 0:
            self._balance += amount
        else:
            raise ValueError("Deposit amount must be positive.")

account = BankAccount("Alex", 1500.0, 1234)
account.deposit(500.0)
print(account.get_balance(1234))  # 2000.0

# Trying to access private variable directly:
# print(account.__pin)  # Raises AttributeError
```

```python
class Vehicle:
    def __init__(self, brand: str, model: str):
        self.brand = brand
        self.model = model

    def start_engine(self):
        return f"{self.brand} {self.model} engine started."

# ElectricCar inherits from Vehicle
class ElectricCar(Vehicle):
    def __init__(self, brand: str, model: str, battery_capacity: int):
        # super() invokes the parent class constructor
        super().__init__(brand, model)
        self.battery_capacity = battery_capacity

    def charge(self):
        return f"Charging the {self.battery_capacity}kWh battery."

car = ElectricCar("Tesla", "Model 3", 75)
print(car.start_engine())  # Inherited from Vehicle
print(car.charge())        # Specific to ElectricCar
```

```python
# Polymorphism 

class Dog:
    def speak(self) -> str:
        return "Woof!"
class Cat:
    def speak(self) -> str:
        return "Meow!"
class Duck:
    def speak(self) -> str:
        return "Quack!"
def make_animal_sound(animal):
    print(animal.speak())

animals = [Dog(), Cat(), Duck()]

for a in animals:
    make_animal_sound(a)

```

```python
# Abstraction 

from abc import ABC, abstractmethod

class PaymentProcessor(ABC):
    @abstractmethod
    def process_payment(self, amount: float) -> None:
        """Every payment gateway implement this"""
        pass

class StripeProcessor(PaymentProcessor):
    def process_payment(self, amount: float) -> None:
        print(f"Authorizing ${amount:.2f} via Stripe API token...")

class PayPalProcessor(PaymentProcessor):
    def process_payment(self, amount: float) -> None:
        print(f"Redirecting user to PayPal to approve ${amount:.2f}...")


gateway: PaymentProcessor = StripeProcessor()
gateway.process_payment(4900.00)
```
