```python
# even or odd 

def check_number(num):
    if num % 2 == 0:
        print(num, "- Even Number")
    else :
        print(num,"- Odd Number")

check_number(21)
```

```python
# factorial of number 

def fact_num(num):
    fact = 1
    
    for i in range(1, num+1):
        fact *= i
    return fact
```

```python
fact_num(5)
```

```python
# sum of natural number till the number provided 

def sum_natural(num):
    total = 0
    for i in range(1, num+1):
        total = total + i
    return total
```

```python
print(sum_natural(10))
```

```python
separate_character = lambda s : (
    [c for c in s if c.lower() in 'aeiou'],
    [c for c in s if c.isalpha() and c.lower() not in 'aeiou']
)

```

```python
vowel, consonants = separate_character("Hello World!")

print("Vowels ", vowel)
print("Consonants :" ,consonants)
```

```python
# Map, Reduce and Filter 


firstname = ["Rahul Vikas kale"]

result = list(map(lambda x : x.upper(), firstname))

print(result)
```

```python
number_list = [1,2,3,4,5,6,7,8,9]

even_numbers = list(filter(lambda i : i % 2 == 0, number_list))

print(even_numbers)
```

```python
from functools import reduce
numbers = [10,20,30,40]

sum_list = reduce(lambda a,b : a + b, numbers)

print(sum_list)
```

```python
import math

def bonus_calculated(salary,bonus):
    employee_salary = salary + bonus
    return employee_salary


employee_salary_record = {
    101 : [12000, 8000],
    102 : [25000, 4000],
    103 : [17000, 5000],
    104 : [10000, 10000],
    105 : [50000, 2000],
}

result = {}

for empId, (i,j) in employee_salary_record.items():
    total_salary = bonus_calculated(i,j)
    result[empId] = total_salary

print(result)
```

```python
from itertools import starmap

def bonus_calculated(salary, bonus):
    return salary + bonus


employee_salary_record = {
    101 : [12000, 8000],
    102 : [25000, 4000],
    103 : [17000, 5000],
    104 : [10000, 10000],
    105 : [50000, 2000],
}

result = list(starmap(bonus_calculated, employee_salary_record.values()))
# result = list(map(lambda pair : bonus_calculated(*pair), employee_salary_record.values()))

totals = map(lambda pair : bonus_calculated(*pair),employee_salary_record.values())
result_dict = dict(zip(employee_salary_record.keys(), totals))
print(result_dict)
print(result)
```
