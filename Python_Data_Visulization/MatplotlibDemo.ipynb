```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
Data = {'Year': [1920,1930,1940,1950,1960, 1970, 1980, 1990, 2000, 2010, 2020],
'Exchange Rate': [65, 69, 71, 64, 62, 59, 72, 71, 75, 78, 81]}

df = pd.DataFrame(Data, columns = ['Year', 'Exchange Rate'])
```

```python
type(Data)
```

```python
df = pd.DataFrame(Data , columns=['Year', 'Exchange Rate'])
```

```python
df
```

```python
df.plot(x = 'Year', y='Exchange Rate', kind='bar', xlabel= 'Years', ylabel='Exchange Rate')
plt.show()
```

```python
df.plot(x = 'Year', y='Exchange Rate', kind='area', xlabel= 'Years', ylabel='Exchange Rate')
plt.show()
```

```python
df.plot(x = 'Year', y='Exchange Rate', kind='line', xlabel= 'Years', ylabel='Exchange Rate')
plt.show()
```

```python
df.plot(x = 'Year', y='Exchange Rate', kind='scatter', xlabel= 'Years', ylabel='Exchange Rate')
plt.show()
```

```python
plt.scatter(df['Year'], df['Exchange Rate'])
plt.show()
```
